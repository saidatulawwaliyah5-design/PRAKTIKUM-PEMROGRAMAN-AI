```python
A = []
B = []
C = []

print('Input Matriks A')
for i in range(2):
    baris = []
    for j in range(2):
        nilai = int(input('A[%d][%d] = ' %(i+1,j+1)))
        baris.append(nilai)
    A.append(baris)

print('Input Matriks B')
for i in range(2):
    baris = []
    for j in range(2):
        nilai = int(input('B[%d][%d] = ' %(i+1,j+1)))
        baris.append(nilai)
    B.append(baris)

for i in range(2):
    baris = []
    for j in range(2):
        baris.append(A[i][j] + B[i][j])
    C.append(baris)

print('Hasil Penjumlahan Matriks:')
for i in range(2):
    for j in range(2):
        print(C[i][j], end=' ')
    print()
```
