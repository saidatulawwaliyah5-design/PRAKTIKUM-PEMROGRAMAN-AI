```python
def perkalian(bil1, bil2):
    hasil = bil1 * bil2
    return hasil

tampil = perkalian(bil2=7, bil1=5)
print("Hasil perkalian:", tampil)
```

```python
nilai_global = 10

def tampil():
    nilai_lokal = 5
    print("Nilai lokal:", nilai_lokal)
    print("Nilai global:", nilai_global)

tampil()
print("Nilai global di luar fungsi:", nilai_global)
```
