```python
class Faktorial:
    def __init__(self, angka):
        self.angka = angka

    def hitung(self):
        hasil = 1
        for i in range(1, self.angka + 1):
            hasil *= i
        return hasil

    def tampilkan(self):
        uraian = "x".join(str(i) for i in range(1, self.angka + 1))
        print(f"{uraian} = {self.hitung()}")


n = int(input("Masukkan angka: "))
f = Faktorial(n)
f.tampilkan()
```

```python
class HitungHuruf:
    def __init__(self, teks):
        self.teks = teks
        self.upper = 0
        self.lower = 0

    def hitung(self):
        for ch in self.teks:
            if ch.isupper():
                self.upper += 1
            elif ch.islower():
                self.lower += 1

    def tampilkan(self):
        self.hitung()
        print("Kalimat:", self.teks)
        print("Jumlah huruf upper case :", self.upper)
        print("Jumlah huruf lower case :", self.lower)


kalimat = input("Masukkan string: ")
h = HitungHuruf(kalimat)
h.tampilkan()
```

```python
class SegitigaPascal:
    def __init__(self, baris):
        self.baris = baris

    def buat(self):
        segitiga = []
        for i in range(self.baris):
            row = [1] * (i + 1)
            for j in range(1, i):
                row[j] = segitiga[i - 1][j - 1] + segitiga[i - 1][j]
            segitiga.append(row)
        return segitiga

    def tampilkan(self):
        segitiga = self.buat()
        for row in segitiga:
            print(" ".join(str(x) for x in row).center(self.baris * 4))


n = int(input("Masukkan jumlah baris: "))
p = SegitigaPascal(n)
p.tampilkan()
```

```python
class KuadratList:
    def __init__(self, daftar):
        self.daftar = daftar

    def hitung(self):
        return [x * x for x in self.daftar]

    def tampilkan(self):
        print("List awal  :", self.daftar)
        print("Hasil kali :", self.hitung())


n = int(input("Masukkan batas angka: "))
k = KuadratList(list(range(1, n + 1)))
k.tampilkan()
```
